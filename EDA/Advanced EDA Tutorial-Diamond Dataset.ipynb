## 💡 What is Advanced EDA?

Advanced EDA is a deeper analysis of data that goes beyond basic exploration to uncover patterns, relationships, distributions, and anomalies. It helps generate stronger insights for feature engineering, hypothesis building, and better machine learning decisions.

#### 📍 Estimate of Location

Variables with measured or count data might have thousands of distinct values. A basic step in exploring the data is getting a “typical value” for each feature (variable): an estimate of where each data is located (i.e., its central tendency).

* __Mean__: The sum of all values divided by the number of values. \
✅ __Best used__: The mean is useful for comparing datasets with similar distributions and no extreme values, such as comparing average income across regions or departments. \
❌ __Not Suitable__: The mean is sensitive to outliers and skewed data because extremely large or small values can significantly affect the result, making it less representative of the dataset's typical value.

* __Trimmed Mean__: The average of all values after dropping a fixed number of extreme values. \
✅ __Best Used__: The trimmed mean is useful for datasets that contain moderate outliers or slight skewness, as it reduces the influence of extreme values while still considering most of the data. \
❌ __Not Suitable__: The trimmed mean may not be appropriate when all data points are important or when the dataset is very small, since removing values can lead to loss of important Information.
* __Median__: The value such that one-half of the data lies above and below. \
✅ __Best Used__: The median is useful for skewed datasets or those with outliers, as it better represents the typical value when the mean can be misleading. \
❌ __Not Suitable__: The median may not be ideal for symmetric datasets when the exact average is needed, as it only represents the middle value and igmores the magnitude of other values.
* __Mode__: Te mode is the most frequent value or category in the dataset. \
✅ __Best Used__: The mode is useful for categorical or discrete data to identify the most frequently ocuuring value, such as the most popular product sold in a store. \
❌ __Not Suitable__: The mode may not be useful for continuous measurements like height or weight that usually do not have a mode.


````python
import numpy as np
from scipy import stats

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    mean_val = df[col].mean()
    median_val = df[col].median()
    trimmed_mean_val = stats.trim_mean(df[col].dropna(), proportiontocut=0.1)
    mode_val = df[col].mode()

    print(f"\nColumn: {col}")
    print("Mean:", mean_val)
    print("Median:", median_val)
    print("Trimmed Mean:", trimmed_mean_val)
    print("Mode:", mode_val)
````

👉 The basic metric for location is the mean, but it can be sensitive to extreme values (outliers). Other metrics (median, trimmed mean) are less sensitive to outliers and unusual distributions and hence are more robust.

#### 🌡️ Estimate of Variability 

Location is just one dimension in summarising a feature. A second dimension, variability, also referred to as dispersion, measures whether the data values are tighty clustered or spread out.

* __Variance__: The sum of squared deviations from the mean divided by n-1 where n is the number of data values. 
* __Standard Deviation__: The square root of the variance. \
✅ __Best Used__: Standard deviation is useful for understanding how spread out the data is. For instance in daily website traffic analysis, a high standard deviation indicates traffic varies greatly from day to day. \
❌ __Not Suitable__: Standard deviation is sensitive to outliers and skewed data, so it may not always represent the variability of the dataset accurately.
* __Median Absolute Deviation (MAD)__: The median of the absolute values of the deviations from the median. \
✅ __Best Used__: The MAD is useful for measuring variability in skewed datasets or datasets with outliers, as it is resistant to extreme values and provides a robust mesaure of spread around the median.
\
❌ __Not Suitable__: The MAD may not be ideal for perfectly symmetric or normally distributed datasets when a more sensitive measure of variability, such as the standard deviation, is needed because MAD does not fully reflect the influence of all data values.
* __Interquartile Range (IQR)__: The Interquartile Range (IQR) is a measure of statistical dispersion calculated as the difference between the third quartile (Q3) and the first quartile (Q1), representing the spread of the middle 50% of the data. \
✅ __Best Used__: IQR helps detect outliers and understand the spread of the middle of the data, such as identifying students who performed significantly better or worse than most of the class when analysing exam scores. \
❌ __Not Suitable__: IQR may be less appropriate for perfectly symmetric distributions without outliers, where measures such as the mean and standard deviation can provide more detailed information about the overall spread of the data.
* __Skewness__: Skewness measures the asymmetry of the data distribution. It indicates whether the data leans toward the right (positive skew) or left (negative skew). \
✅ __Best Used__: Skewness helps determine if data transformation is needed when the distribution is highly skewed. \
❌ __Not Suitable__: Calculating skewness for symmetric data is not necessary as it will be close to zero.
* __Kurosis__: Kurtosis measures the tailedness of a distribution, indicating whetger data has heavy or light tails compared to a normal distribution. High kurtosis suggests more extreme outliers, while low kurtosis indicates fewer extreme values. \
✅ __Best Used__: High kurtosis might signal that we need to pay more attention to outliers and extreme values. \
❌ __Not Suitable__: If a dataset is already fairly well-behaved with a near-normal distribution, kurtosis might not provide additional information.

💡 The standard deviation is much easier to interpret than the variance since it is on the same scale as the original data.
Neither the variance, nor the standard deviation is robust to outliers and extreme values. A robust estimate of variability is the MAD. Like the median, the MAD is not influenced by extreme values.

````python
from scipy.stats import skew, kurtosis

for col in numeric_cols:
    variance = np.var(df[col])          # Population variance
    sample_variance = np.var(df[col], ddof=1)  # Sample variance
    std_dev = df[col].std()
    IQR = stats.iqr(df[col])     # Interquartile Range (IQR)
    MAD = stats.median_abs_deviation(df[col])     # Median Absolute Deviation (MAD)
    skewness = stats.skew(df[col])
    kurtosis = stats.kurtosis(df[col])

    print(f"\nColumn: {col}")
    print("Variance:", variance)
    print("Sample_Variance:", sample_variance)
    print("std_dev:", std_dev)
    print("IQR:", IQR)
    print("MAD:", MAD)
    print("Skewness:", skewness)
    print("Kurtosis:", kurtosis)
````

Data quality assessment identified 146 duplicate records, representing a very small proportion of the dataset. The average diamond weighs approximately 0.80 carats, although the median of 0.70 carats indicates that most diamonds are relatively small, with a few larger diamonds pulling the mean upward. This pattern is confirmed by the positive skewness of the carat distribution (1.11). Price exhibits even stronger right-skewness (1.62), with an average value of 3,933 compared to a median of $2,401, suggesting the presence of a relatively small number of high-value diamonds. The physical dimensions (x, y, and z) show moderate variability, although the y and z measurements exhibit extreme kurtosis values (91.74 and 47.38 respectively), indicating the presence of significant outliers or unusual observations.

The depth variable is relatively stable, with a mean of 61.75 and a near-symmetric distribution (skewness = -0.11), while the table variable shows moderate positive skewness (0.79). Variability measures such as variance, standard deviation, IQR, and MAD suggest substantial spread in price and carat compared to the other features. Several numerical variables display non-normal distributions, particularly price, carat, y, and z, which contain heavy tails and extreme values. These findings suggest that transformations such as logarithmic scaling may be beneficial for modelling. Overall, the dataset demonstrates strong variability in diamond size and price, contains a small number of duplicates, exhibits several highly skewed and heavy-tailed distributions, and provides rich opportunities for exploratory analysis, feature engineering, and predictive modelling of diamond prices.

#### __2. Distributions__

Histograms and density plots provide the most intuitive visualizations of a distribution, but both require arbitrary parameter choices and can be misleading. Cumulative densities and quantile-quantile (q-q) plots always represent the data faithfully but can be more difficult to interpret.

Boxplots, violin plots, strip charts, and sina plots are useful when we want to visualize many distributions at once and/or if we are primarily interested in overall shifts among the distributions. Stacked histograms and overlapping densities allow a more in-depth comparison of a smaller number of distributions, though stacked histograms can be difficult to interpret and are best avoided. Ridgeline plots can be a useful alternative to violin plots and are often useful when visualizing very large numbers of distributions or changes in distributions over time.

#### __3. Proportions__

Proportions can be visualized as pie charts, side-by-side bars, or stacked bars. As for amounts, when we visualize proportions with bars, the bars can be arranged either vertically or horizontally. Pie charts emphasize that the individual parts add up to a whole and highlight simple fractions. However, the individual pieces are more easily compared in side-by-side bars. Stacked bars look awkward for a single set of proportions, but can be useful when comparing multiple sets of proportions.

<p align="center">
    <img src="image5.PNG" width="800">
</p>

When visualizing multiple sets of proportions or changes in proportions across conditions, pie charts tend to be space-inefficient and often obscure relationships. Grouped bars work well as long as the number of conditions compared is moderate, and stacked bars can work for large numbers of conditions. Stacked densities are appropriate when the proportions change along a continuous variable.

When proportions are specified according to multiple grouping variables, mosaic plots, treemaps, or parallel sets are useful visualization approaches. Mosaic plots assume that every level of one grouping variable can be combined with every level of another grouping variable, whereas treemaps do not make such an assumption. Treemaps work well even if the subdivisions of one group are entirely distinct from the subdivisions of another. Parallel sets work better than either mosaic plots or treemaps when there are more than two grouping variables.

#### 📊 __Bar Plot__

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

for col in cat_cols:

    plt.figure(figsize=(8, 4))

    ax = sns.countplot(
        data=df,
        x=col, 
        color="skyblue",
        order=df[col].value_counts().index
    )

    for container in ax.containers:
        ax.bar_label(container, fmt='%d', padding=3, fontsize=9)

    plt.title(f'Distribution of {col}', fontsize=14, weight='bold')
    plt.xlabel("Count", fontsize=11)
    plt.ylabel(col, fontsize=11)

    sns.despine()
    plt.tight_layout()
    plt.show()
````

🧠 The count plots of the categorical variables provide valuable insight into the composition of the Diamonds dataset. The distribution of diamond cuts reveals that the majority of diamonds are classified as Ideal or Premium, while Fair diamonds represent only a small fraction of the dataset. Similarly, the colour distribution shows that mid-range color grades such as G, E, and F are the most common, whereas the highest and lowest color grades appear less frequently. The clarity distribution follows a similar pattern, with SI1, VS2, and SI2 dominating the dataset while the rarest clarity grades occur in much smaller numbers. These visualisations highlight a moderate class imbalance across categorical features, indicating that some diamond quality categories are substantially more represented than others. From a business perspective, this suggests that the dataset primarily reflects mainstream market inventory rather than rare premium diamonds. Overall, the plots demonstrate that the dataset contains a diverse range of diamond qualities, with certain cut, colour, and clarity categories dominating the market, providing valuable context for subsequent price analysis, customer segmentation, and predictive modeling.

#### 🌲 __Treemap__

````python
import squarify

cut_counts = df['cut'].value_counts()

colors = sns.color_palette("Greens", len(cut_counts))

plt.figure(figsize=(8, 5))

squarify.plot(
    sizes=cut_counts.values,
    label=cut_counts.index,
    color=colors,
    alpha=0.9
)

plt.title("Treemap of Diamond Cuts", fontsize=18)
plt.axis('off')

plt.show()
````

🧠 The treemap visualisation of diamond cuts provides a clear overview of the structural composition of the Diamonds dataset. #### 🥵 __Categorical Relationship Heatmap__

Seaborn’s heatmap is designed for numerical correlations. However, you can analyse associations between categorical variables by using statistical measures such as **Cramér’s V** or **Theil’s U** instead of standard correlation coefficients. Here’s how to compute Cramér’s V for all pairs of categorical variables and plot the results in a heatmap.

````python
import plotly.express as px
from scipy.stats import chi2_contingency


def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()

    phi2 = chi2 / n
    r, k = confusion_matrix.shape

    phi2corr = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1))
    rcorr = r - ((r - 1) ** 2) / (n - 1)
    kcorr = k - ((k - 1) ** 2) / (n - 1)

    return np.sqrt(phi2corr / min((kcorr - 1), (rcorr - 1)))


cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

cramers_results = pd.DataFrame(index=cat_cols, columns=cat_cols)

for col1 in cat_cols:
    for col2 in cat_cols:
        confusion_mat = pd.crosstab(df[col1], df[col2])
        cramers_results.loc[col1, col2] = cramers_v(confusion_mat)

cramers_results = cramers_results.astype(float)


cramers_long = cramers_results.reset_index().melt(id_vars='index')
cramers_long.columns = ['Variable 1', 'Variable 2', "Cramér's V"]


fig = px.imshow(
    cramers_results,
    text_auto=".2f",
    color_continuous_scale="GnBu",
    zmin=0, zmax=1,
    aspect="auto",
    title="Interactive Cramér’s V Heatmap (Categorical Associations)"
)

fig.update_layout(
    width=800,
    height=700,
    title_x=0.5
)

fig.show()
````

🧠 
The heatmap shows the strength of association between pairs of categorical features on a scale from 0 (no relationship) to 1 (strong relationship). From the visualisation, we typically observe that cut, colour, and clarity are moderately related, which is expected because diamond grading characteristics are not completely independent. However, no perfect correlations appear, indicating that each feature still contributes unique information. Overall, this plot helps us understand the dependency structure among categorical variables, which is useful before building predictive models or performing feature engineering.

### 7️⃣ Outlier Detection

Outlier detection is a crucial step in exploratory data analysis because it helps identify unusual or extreme values that can distort statistical summaries and negatively impact machine learning models. One of the most common visual tools for detecting outliers is the **boxplot**, which highlights potential extreme values based on the spread of the data using quartiles and whiskers. From a statistical perspective, the **IQR (Interquartile Range)** method defines outliers as observations that fall below Q1 − 1.5×IQR or above Q3 + 1.5×IQR, making it a robust and widely used rule. Another approach is the **Z-score** method, which measures how many standard deviations a data point is away from the mean, typically flagging values with a Z-score greater than ±3 as outliers in normally distributed data. More advanced machine learning techniques include **Local Outlier Factor (LOF)**, which identifies anomalies by comparing the local density of a data point to the densities of its neighboring points, making it effective for detecting local deviations in complex datasets. Similarly, **Isolation Forest** detects outliers by randomly partitioning the data and isolating observations; anomalous points tend to require fewer splits to separate from the rest of the dataset, making this method highly efficient for large and high-dimensional data. Finally, **domain-based filtering** incorporates real-world knowledge to identify invalid or implausible values that may not be statistically extreme but are logically incorrect (for example, a negative carat value in diamonds). Together, these methods provide complementary statistical, machine learning, and domain-driven perspectives for detecting and handling outliers effectively.


The three most common ways of handling outliers are: 
1. 🗑️ __Removing Outliers__ \
Extreme values are completely removed from the dataset. \
\
Common methods: 
* IQR method 📏
* Z-score method 📉 \
✅ Useful when outliers are: 
* data errors
* noise
* unrealistic observations \
⚠️ Risk: \
important information may be lost 
2. 🔄 __Transforming Outliers__ \
Mathematical transformations are applied to reduce the impact of extreme values. \
\
Common transformations:

* square root transform
* Box-Cox transform \
✅ Useful for:

3. ✂️ __Capping (Winsorization)__ \
Instead of deleting outliers, extreme values are replaced with upper or lower threshold values. \
\
Example: replace values above the 99th percentile with the 99th percentile
* ✅ Preserves dataset size
* ✅ Reduces influence of extreme values

#### 📏 __IQR Method__

The Interquartile Range (IQR) method is a statistical technique used to detect outliers based on the spread of the middle 50% of the data. It calculates the range between the first quartile (Q1) and third quartile (Q3), then defines outliers as values falling below Q1 − 1.5×IQR or above Q3 + 1.5×IQR. Because it relies on quartiles instead of the mean, the IQR method is robust to skewed distributions and extreme values.

````python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)

IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
df_no_outliers = df[
    (df['price'] >= lower_bound) &
    (df['price'] <= upper_bound)
]

print("Original dataset shape:", df.shape)
print("Dataset shape after removing outliers:", df_no_outliers.shape)
````

````python
plt.figure(figsize=(6, 4))

sns.boxplot(
    data=df_no_outliers,
    y='price',
    color='#00a896'
)

plt.title('Diamond Prices After Outlier Removal', fontsize=15, weight='bold')
plt.ylabel('Price')

plt.tight_layout()
plt.show()
````

- The boxplot after IQR-based outlier removal shows a much more compact and balanced distribution of diamond prices, with the extreme high-value observations successfully excluded.
- Removing these outliers reduces the strong right-skewness previously present in the price variable, allowing the central price distribution to be visualised more clearly and making the dataset more suitable for statistical analysis and predictive modelling.

The Z-score method identifies outliers by measuring how far a data point is from the mean in terms of standard deviations. A Z-score indicates the relative position of a value within the distribution, and observations with scores greater than ±3 are commonly considered outliers. This method works best for approximately normal distributions and is useful for detecting unusually extreme observations statistically.
\
\
✅ __Best Used__: This method is mostly useful when dealing with normaly distributed data, as they help quantify how far each point is from the mean.
\
❌ __Not Suitable__: Z-scores are less useful when the data is not normally distributed, as they rely on the assumption that data follows a bell-shaped curve.

#### 🌳 __Isolation Forest__

Isolation Forest is an anomaly detection algorithm designed to find unusual data points (outliers).
\
The main idea is simple: 
\
__Outliers are easier to isolate than normal points.__
\
Imagine you randomly split data again and again like cutting branches in a tree.
* Normal data points are surrounded by many similar points → harder to isolate.
* Outliers are far away from others → isolated very quickly.


How it works step by step: 
1. The algorithm randomly selects a feature.
2.	It randomly selects a split value.
3.	Data gets divided into smaller groups.
4.	This process continues recursively, creating many random trees.

If a point gets isolated after only a few splits: 
* it is probably an outlier.

If it needs many splits: 
* it is likely normal.

✅ __Best Used__: Isolation Forest works best for large datasets, high-dimensional data, and problems where anomalies follow complex patterns that are difficult to detect with simple statistical methods. It is fast, scalable, and efficient even when working with many features. Another major advantage is that it does not require the data to follow a normal distribution, making it suitable for real-world datasets with irregular structures.



__❌ Not Suitable__: However, Isolation Forest may not perform well on very small datasets because the random partitioning process can become unstable. It can also struggle when anomalies are not clearly separated from normal observations or when the dataset contains a high level of noise. In addition, because the algorithm relies on randomness, the detected anomalies may vary slightly between different runs.


````python
from sklearn.ensemble import IsolationForest
from scipy.stats.mstats import winsorize

num_cols = ['carat', 'depth', 'table', 'price', 'x', 'y', 'z']
X = df[num_cols]

# Create Isolation Forest model
iso = IsolationForest(
    contamination=0.02,
    random_state=42
)


df['outlier'] = iso.fit_predict(X)


n_outliers = (df['outlier'] == -1).sum()

print("Dataset Shape:", df.shape)
print("Number of Outliers Detected:", n_outliers)

df_wins = df.copy()

# Winsorize numerical columns
for col in num_cols:
    df_wins[col] = winsorize(
        df_wins[col],
        limits=[0.02, 0.02]  # Cap lowest 2% and highest 2%
    )

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['price'])
plt.title("Before Winsorization")

plt.subplot(1, 2, 2)
sns.boxplot(y=df_wins['price'])
plt.title("After Winsorization")

plt.tight_layout()
plt.show()
````

- Outliers in the diamonds dataset were identified using an **Isolation Forest model**, which is well-suited for detecting anomalies in high-dimensional numerical data.
- The model was applied to key numerical features including **carat, depth, table, price, and diamond dimensions (x, y, z)**, with a contamination rate of **2%**, meaning the algorithm assumes approximately 2% of the data may be anomalous.

- The model flagged a subset of observations as outliers (`outlier = -1`), highlighting diamonds with unusual combinations of size, depth, or price compared to the rest of the dataset.
- These detected outliers typically correspond to **extreme values in price and physical dimensions**, which can heavily influence statistical summaries and machine learning models.

- To reduce the impact of extreme values, **winsorization was applied**, capping both the lower and upper **2% of values** in each numerical feature.
- This approach preserves all observations but limits the influence of extreme tails by bringing extreme values closer to the main distribution.

- The comparison of boxplots before and after winsorization shows a **significant reduction in extreme values**, especially in the price distribution.
- After treatment, the dataset becomes more stable and less skewed, improving its suitability for modelling and statistical analysis.

- Overall, combining **Isolation Forest for detection** and **winsorization for treatment** provides a robust approach to handling outliers while maintaining the integrity of the dataset.

#### 🏭 __LOF (Local Outlier Factor)__

LOF detects anomalies by comparing the local density of a point to the density of its neighbours.
\
The intuition:
\
Outliers live in sparse regions.
\
If a point has much lower density than nearby points, it is likely abnormal.
\
\
__How LOF works__

__Step 1__ — Find nearest neighbors
\
For every point:
* find its k-nearest neighbors.




__Step 2__— Compute local density
\
Density means:
* how crowded the area around the point is.

Dense area → normal
\
Sparse area → suspicious
\
__Step 3__ — Compare densities
\
__LOF compares:__
* density of the point
\
vs
* density of its neighbors.

If the point is significantly less dense:
* LOF score becomes high
* point is labeled an outlier.


✅ __Best Used__: LOF works best for detecting anomalies in datasets where the local density of data points is important. It is particularly effective when outliers are surrounded by much denser neighborhoods than normal observations. LOF performs well on datasets with complex structures because it compares each point to its nearby neighbors rather than relying on global assumptions. This makes it useful for identifying subtle or context-dependent anomalies that other methods may miss.

❌ __Not Suitable__: However, LOF can become computationally expensive on very large datasets because it relies heavily on distance calculations between neighboring points. It may also struggle in high-dimensional data where distance measures become less meaningful. In addition, the performance of LOF is sensitive to the choice of the number of neighbors (k), and selecting an inappropriate value can lead to inaccurate outlier detection. Noise and overlapping clusters can also reduce its effectiveness.

````python
from sklearn.neighbors import LocalOutlierFactor

num_cols = ['carat', 'depth', 'table', 'price', 'x', 'y', 'z']
X = df[num_cols]

# Create LOF model
lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.02
)

df['outlier'] = lof.fit_predict(X)

n_outliers = (df['outlier'] == -1).sum()

print("Dataset Shape:", df.shape)
print("Number of Outliers Detected:", n_outliers)

# Apply log transformation to numerical features
df_log = df.copy()

for col in num_cols:
    df_log[col] = np.log1p(df_log[col])

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['price'])
plt.title("Before Log Transformation")

plt.subplot(1, 2, 2)
sns.boxplot(y=df_log['price'])
plt.title("After Log Transformation")

plt.tight_layout()
plt.show()
````

#### 📉 __Dimensionality Reduction__

Dimensionality reduction is a feature engineering technique used to reduce the number of input variables by transforming high-dimensional data into a lower-dimensional representation while preserving as much important information as possible. Methods such as **Principal Component Analysis (PCA)**, **t-SNE**, and **UMAP** create new composite features that capture the most significant patterns and variance in the data. This helps simplify complex datasets while retaining their underlying structure. 

#### 🗜️ **Principal Component Analysis (PCA)**

Principal Component Analysis (PCA) is one of the most widely used dimensionality reduction techniques. It transforms a set of possibly correlated features into a smaller set of uncorrelated variables called principal components. These components are ordered so that the first few capture most of the variation present in the original dataset. By retaining only the most informative components, PCA reduces the complexity of the data while preserving its essential patterns and structure.\
✅ Best Used: PCA is useful for high-dimensional datasets with many correlated features. It helps reduce noise, improve computational efficiency, mitigate multicollinearity, and visualize complex datasets in two or three dimensions while retaining most of the original information.\
❌ Not Suitable: When feature interpretability is important, since principal components are combinations of multiple original features and may not have a clear real-world meaning. PCA is also less effective when the underlying relationships in the data are highly non-linear, as it captures only linear patterns.

````python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = df[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']]

# Standardise the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca,
    columns=['PC1', 'PC2']
)

print(pca_df.head())

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print(f"\nTotal Variance Explained: "
      f"{pca.explained_variance_ratio_.sum():.2%}")
````

````python
plt.figure(figsize=(8, 6))
plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    alpha=0.4
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of Diamonds Dataset')
plt.show()
````

````python
from sklearn.decomposition import PCA

pca_full = PCA()
pca_full.fit(X_scaled)

cum_var = np.cumsum(pca_full.explained_variance_ratio_)

# Components required for 95% variance
n_components = np.argmax(cum_var >= 0.95) + 1

print(f"Components needed for 95% variance: {n_components}")
````

🌀 **t-Distributed Stochastic Neighbour Embedding (t-SNE)**

t-Distributed Stochastic Neighbour Embedding (t-SNE) is a non-linear dimensionality reduction technique designed primarily for data visualization. Unlike PCA, which focuses on preserving overall variance, t-SNE aims to preserve the local structure of the data by keeping similar observations close together in the lower-dimensional space. It converts similarities between data points into probabilities and then maps the data into two or three dimensions, making clusters, patterns, and hidden relationships easier to visualize.

✅ **Best Used:** t-SNE is useful for exploring high-dimensional datasets and visualizing complex non-linear structures. It excels at revealing clusters, groups, and hidden patterns that may not be visible using linear techniques such as PCA. It is commonly used for visualizing image data, text embeddings, biological data, and machine learning feature spaces.

❌ **Not Suitable:** When the goal is feature reduction for predictive modeling or when preserving global distances and relationships is important. t-SNE can be computationally expensive on large datasets, and the distances between clusters in the visualization do not necessarily reflect true distances in the original feature space. Additionally, the results can vary depending on hyperparameters such as perplexity and random initialization.


````python
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE

X = df[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

sample_size = 5000

X_sample = X_scaled[:sample_size]
color_sample = df['cut'][:sample_size]

tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate='auto',
    random_state=42
)

X_tsne = tsne.fit_transform(X_sample)

tsne_df = pd.DataFrame(
    X_tsne,
    columns=['tSNE1', 'tSNE2']
)

tsne_df['cut'] = color_sample.values

plt.figure(figsize=(10, 8))

sns.scatterplot(
    data=tsne_df,
    x='tSNE1',
    y='tSNE2',
    hue='cut',
    alpha=0.7
)

plt.title('t-SNE Visualization of Diamonds Dataset')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.legend(title='Cut')
plt.show()
````

🌐 **Uniform Manifold Approximation and Projection (UMAP)**

Uniform Manifold Approximation and Projection (UMAP) is a non-linear dimensionality reduction technique used for visualizing high-dimensional data in a lower-dimensional space while preserving both local and global structure more effectively than many alternatives. In the context of the **Diamonds dataset**, UMAP projects features such as *carat, price, depth, table, and physical dimensions (x, y, z)* into 2D space, allowing complex relationships between diamonds to be visualized as meaningful clusters and gradients. It is built on manifold learning and topological assumptions, making it particularly powerful for uncovering the intrinsic structure of data.

✅ **Best Used:** UMAP is highly effective for large, high-dimensional datasets like the Diamonds dataset where the goal is to explore clusters, patterns, and groupings in both local neighbourhoods and overall data structure. It is faster than t-SNE, scales better to large datasets, and often produces more globally meaningful visualisations while still preserving fine-grained relationships. It is widely used in exploratory data analysis, image embeddings, and biological or financial datasets.

❌ **Not Suitable:** When strict mathematical interpretability or exact distance preservation is required, or when the primary goal is feature engineering for predictive modeling. Although UMAP preserves more global structure than t-SNE, the transformed axes are still not directly interpretable, and results can vary depending on parameters such as `n_neighbors` and `min_dist`.


````python
from sklearn.preprocessing import StandardScaler
import umap.umap_ as umap

X = df[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']]

# Standardize features (important for UMAP)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_sample = X_scaled[:10000]
color_sample = df['cut'][:10000]

reducer = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap = reducer.fit_transform(X_sample)

umap_df = pd.DataFrame(X_umap, columns=['UMAP1', 'UMAP2'])
umap_df['cut'] = color_sample.values

plt.figure(figsize=(10, 8))

sns.scatterplot(
    data=umap_df,
    x='UMAP1',
    y='UMAP2',
    hue='cut',
    alpha=0.7
)

plt.title("UMAP Visualisation of Diamonds Dataset")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.legend(title="Cut")
plt.show()
````

### 9️⃣ Distribution Checks & Assumptions



Before building any statistical or machine learning model, it is important to understand the structure and distributional properties of the data. **Normality tests** such as the **Shapiro–Wilk test**, along with **skewness measures**, help assess whether a variable follows an approximately normal distribution or is significantly asymmetric. This is crucial because many statistical methods assume normality, and violations can affect model performance and inference. In parallel, checking for **class imbalance** ensures that categorical outcomes are not disproportionately represented, which could bias a model toward majority classes and lead to misleading accuracy. Additionally, detecting **heavy tails** helps identify whether extreme values occur more frequently than expected under a normal distribution, which can impact variance and robustness of models. Finally, **identifying zero inflation**—where an excessive number of observations are zero—reveals structural patterns in the data that may require specialised modeling approaches rather than standard regression or classification techniques.


Here’s a clean and practical Python implementation for the diamonds dataset covering:
* Normality tests (Shapiro-Wilk)
* Skewness check
* Class imbalance check (for categorical target like cut/color/clarity)
* Heavy tails detection (kurtosis)
* Zero inflation detection

````python
from scipy import stats

# Normality test (Shapiro-Wilk)
numeric_cols = df.select_dtypes(include=np.number).columns

print("=== Shapiro-Wilk Normality Test ===")
for col in numeric_cols:
    sample = df[col].dropna().sample(5000, random_state=42) 
    stat, p = stats.shapiro(sample)
    print(f"{col}: statistic={stat:.4f}, p-value={p:.4e}")
    if p < 0.05:
        print(" -> Not normally distributed\n")
    else:
        print(" -> Approximately normal\n")
````
